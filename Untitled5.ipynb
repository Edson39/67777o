## Como conectar seu notebook do Colab ao GitHub

1.  **Salvar uma cópia no GitHub:**
    *   No Colab, vá para `File` (Arquivo) -> `Save a copy in GitHub` (Salvar uma cópia no GitHub).
    *   Você será solicitado a autorizar o Colab a acessar sua conta do GitHub, se ainda não o fez.
    *   Escolha o repositório, o nome do arquivo e adicione uma mensagem de commit.
    *   Clique em `OK`.

2.  **Abrir um notebook do GitHub no Colab:**
    *   No Colab, vá para `File` (Arquivo) -> `Open notebook` (Abrir notebook).
    *   Clique na aba `GitHub`.
    *   Você pode pesquisar por repositórios ou colar o URL de um notebook do GitHub.
    *   Selecione o notebook que deseja abrir.

3.  **Habilitar o controle de versão:**
    *   Para ter um controle de versão mais robusto, você pode clonar um repositório GitHub para o ambiente de tempo de execução do Colab usando comandos `git` no notebook.
    *   Primeiro, você precisará gerar um Personal Access Token (PAT) no GitHub com as permissões necessárias (geralmente `repo`).
    *   Você pode usar este token para clonar seu repositório:
    ```bash
    !git clone https://<YOUR_GITHUB_TOKEN>@github.com/<YOUR_USERNAME>/<YOUR_REPOSITORY_NAME>.git
    ```
    *   Navegue para o diretório do seu repositório clonado e trabalhe nos seus arquivos.
    *   Para salvar as alterações de volta ao GitHub, use os comandos `git add`, `git commit` e `git push`.
